```python
#verificando o schema
df_bronze = spark.table("mercado_financeiro.bronze.ibvespa_acoes")
df_bronze.printSchema()
```

```python
#visualizando a primeiras linhas da camada bronz (como ficou após salvarmos)
df_bronze.show(5)
```

```python
#INICIO DA CAMADA SILVER - Copiando a estrutura da tabela bronze 

df_silver = df_bronze
```

```python
#Visualizando a estrutura da tabela silver
df_silver.show(5)
```

```python
#Tratamento da coluna Ticker - removendo o .SA e deixando somente o código do ticker
#Criando uma coluna nova com o ticker corrigido

from pyspark.sql.functions import regexp_replace, split, substring, length, col
df_silver = df_silver.withColumn(
    "Ticker",
    regexp_replace(col("Ticker"), "\\.SA", "")
)
```

```python
#após o tratamento da coluna ticker vamos ver como ficou
df_silver.show(5)
```

```python
#renomear as colunas

df_silver = (
    df_silver
    .withColumnRenamed('Date', 'Data')
    .withColumnRenamed('Ticker', 'Codigo')
    .withColumnRenamed('Open', 'Abertura')
    .withColumnRenamed('High', 'Maximo')
    .withColumnRenamed('Low', 'Minimo')
    .withColumnRenamed('Close', 'Fechamento')
    .withColumnRenamed('Adj_Close', 'Fechamento_Ajustado')
   
)
```

```python
#visualizando após renomear as colunas 

df_silver.show(5)
```

```python
df_silver.printSchema()
```

```python
#vamos salvar a tabela 
(
df_silver.write
  .format('delta')
  .mode('overwrite')
  .saveAsTable('mercado_financeiro.silver.ibvespa_acoes')

)
```
